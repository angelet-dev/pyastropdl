```python
import pyastropdl as pdl
import timeit
```

```python
# ==============================================================================
# LUNAR SOFT LANDING - REFERENCE SCENARIOS (Static gravitation)
# ==============================================================================

reference_scenarios = {
    "standard_llo_descent": {
        "description": "Standard LLO Descent (TWR ~1.42)",
        "init_cond": [28877, -56, 102378],  # [m_0, v_0, h_0]
        "end_cond": [0, 0],                 # [v_tau, h_tau]
        "args": (66628, 0.00029903701260349804, 17644)  # (u_max, k, m_fuel)
    },

    "heavy_cargo_low_twr": {
        "description": "Heavy Cargo Landing (TWR ~1.04, Critical Mass)",
        "init_cond": [38852, -109, 105674],
        "end_cond": [0, 0],
        "args": (65848, 0.0002832545036049801, 15358)
    },

    "terminal_burn": {
        "description": "Powered Descent Initiation (High TWR ~9.0)",
        "init_cond": [5712, -228, 13643],
        "end_cond": [0, 0],
        "args": (83388, 0.00028246986509083884, 1594)
    },

    "hoverslam": {
        "description": "Hoverslam / High-G Terminal Burn",
        "init_cond": [3885, -482, 203096],
        "end_cond": [0, 0],
        "args": (68645, 0.0003137588347624395, 2355)
    },

    "orbital_drop": {
        "description": "Orbital Drop (Zero Initial Radial Velocity)",
        "init_cond": [46535, 4, 106691],
        "end_cond": [0, 0],
        "args": (98004, 0.0003892046614419574, 27900)
    }
}


for case_name, params in reference_scenarios.items():
    print(f"\n[ EXECUTING ]: {params['description']}")
    
    solver = pdl.Solver(params["init_cond"], params["end_cond"], params["args"], sim_engine='static_g')
    
  
    exec_time = timeit.timeit(lambda: solver.solve(dt=0.01), number=1)
    print(f"[ SUCCESS ]: Solved in {exec_time:.4f} seconds.")

    exec_time = timeit.timeit(lambda: solver.find_residuals(dt=0.1), number=1)      
    print(f"[ SUCCESS ]: Find in {exec_time:.4f} seconds.")

    solver.print_results()
    
    plotter = pdl.Plotter(solver)
    plotter.plot_trajectory(f"results/plots/st_trj_{case_name}.png")
    plotter.plot_telemetry(f"results/plots/st_ts_{case_name}.png")
    plotter.plot_residuals(f"results/plots/st_res_{case_name}.png")

    # Animate the phase trajectory
    # animator = pdl.Animator(solver)
    # animator.animate(60, 15, f"results/video/st_ani_{case_name}.mp4")
```
